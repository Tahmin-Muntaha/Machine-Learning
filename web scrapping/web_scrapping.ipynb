```python
ol=soup.find('ol',class_='row')
```

```python
name=[]
price=[]
rating=[]

for i in ol.find_all('li'):
    h3=i.find('h3')
    name.append((h3.find('a').text.strip()))
    p=i.find('p',class_='price_color').text
    price.append(p)
    r=i.find('p',class_='star-rating')
    rating.append(r['class'][1])
    # extract values from class attribute .soup mainly read attributes as list
d={'name':name,'price':price,'rating':rating}
    
```

```python
final=pd.DataFrame()


```

```python
for j in range(1,51):
    name=[]
    price=[]
    rating=[]
    webpage=requests.get(f'https://books.toscrape.com/catalogue/page-{j}.html').text
    soup=BeautifulSoup(webpage,'lxml')
    ol=soup.find('ol',class_='row')
    
    for i in ol.find_all('li'):
        h3=i.find('h3')
        name.append((h3.find('a').text.strip()))
        p=i.find('p',class_='price_color').text
        price.append(p)
        r=i.find('p',class_='star-rating')
        rating.append(r['class'][1])
        # extract values from class attribute .soup mainly read attributes as list
    d={'name':name,'price':price,'rating':rating}
    x=pd.DataFrame(d)
    final=pd.concat([final,x],ignore_index=True)
print(final)
```

```python
final.to_csv('books.csv')
```
