```python
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
import numpy as np
```

```python
X,y=load_diabetes(return_X_y=True)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=4)
```

```python
from sklearn.linear_model import SGDRegressor
```

```python
sdg=SGDRegressor(penalty='l2',max_iter=500,eta0=0.1,learning_rate='constant',alpha=0.001)
```

```python
sdg.fit(X_train,y_train)
print(r2_score(y_test,sdg.predict(X_test)))
print(sdg.coef_)
print(sdg.intercept_)
```

```python
from sklearn.linear_model import Ridge
reg=Ridge(alpha=0.001,max_iter=500,solver='sparse_cg')
```

```python
reg.fit(X_train,y_train)
print(r2_score(y_test,reg.predict(X_test)))
print(reg.coef_)
print(reg.intercept_)
```

```python
class AmrGD:
    def __init__(self,epoch,n,alpha):
        self.n=n
        self.epoch=epoch
        self.alpha=alpha
        self.coef=None
        self.inter=None
    def fit(self,X_train,y_train):
        self.coef=np.ones(X_train.shape[1])
        self.inter=0
        w=np.insert(self.coef,0,self.inter)
        X_train=np.insert(X_train,0,1,axis=1)
        for i in range(self.epoch):
            w_der=np.dot(X_train.T,X_train).dot(w) - np.dot(X_train.T,y_train) + self.alpha*w
            w=w-(self.n*w_der)
        
        self.inter=w[0]
        self.coef=w[1:]
        print(self.inter,self.coef)
        
    def predict(self,X_test):
        return np.dot(X_test,self.coef)+self.inter
```

```python
amr=AmrGD(epoch=500,n=0.005,alpha=0.001)
```

```python
amr.fit(X_train,y_train)
```

```python
r2_score(y_test,amr.predict(X_test))
```
