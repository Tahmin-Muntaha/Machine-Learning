```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
X,y=make_regression(n_samples=4,n_features=1,n_informative=1,n_targets=1,noise=80,random_state=13)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
```

```python
lr.fit(X,y)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red')
```

```python
#lets apply gradiant descent assuming slope is constant m=78.35
#and lets assume the starting value for intercept b=0
```

```python
y_pred= ((78.35*X)+0) #y_pred=mx+b
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red')
plt.plot(X,y_pred,color='blue')
```

```python
m=78.35
b=0
```

```python
loss_slope=-2*(np.sum(y-m*X.ravel()-b)) # calculate using the formula
```

```python
loss_slope
```

```python
n=0.1 #learning rate
```

```python
#calculate the new intercept
#iteration 1
```

```python
b=b-(n*loss_slope)
```

```python
b
```

```python
y_pred1= ((78.35*X)+b) #y_pred=mx+b
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red')
plt.plot(X,y_pred,color='green')
plt.plot(X,y_pred1,color='blue')
```

```python
b=-100
m=78.35
n=0.1
epoch=10
for i in range(epoch):
    slope=-2*np.sum(y-m*X.ravel()-b)
    b=b-(n*slope)
    y_pred=m*X+b
    plt.plot(X,y_pred)
plt.scatter(X,y)  
```
