```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('train.csv',usecols=['Age','Pclass','SibSp','Parch','Survived'])
```

```python
df
```

```python
df.dropna(inplace=True)
```

```python
df
```

```python
#feature construction
df['Family_size']=df['SibSp']+df['Parch']+1
```

```python
df
```

```python
def func(num):
    if num==1:
        return 0
        #alone
    elif num>1 and num<=4:
        return 1
        #small family
    else:
        return 2
        #large family

```

```python
df['Family_type']=df['Family_size'].apply(func)
```
