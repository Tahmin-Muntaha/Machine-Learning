```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('train.csv',usecols=['Survived','Age','SibSp','Parch','Fare'])
```

```python
df.head()
```

```python
df['family']=df['SibSp']+df['Parch']
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df
```

```python
df.drop(columns=['SibSp','Parch'],inplace=True)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df.iloc[:,1:],df.iloc[:,0],test_size=0.2,random_state=False)
```

```python
from sklearn.preprocessing import Binarizer
from sklearn.compose import ColumnTransformer
```

```python
trf=ColumnTransformer([
    ('bin',Binarizer(copy=False),['family'])
],remainder='passthrough')
```

```python
X_train_trf=trf.fit_transform(X_train)
X_test_trf=trf.transform(X_test)
```

```python
X_train_final=pd.DataFrame(X_train_trf,columns=trf.get_feature_names_out())
```
