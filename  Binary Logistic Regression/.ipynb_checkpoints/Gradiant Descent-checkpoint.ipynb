```python
from sklearn.datasets import make_classification
import numpy as np
```

```python
X,y=make_classification(n_samples=100,n_features=2,n_informative=1,n_redundant=0,
                       n_classes=2,n_clusters_per_class=1,random_state=41,hypercube=False,
                       class_sep=20)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X[:,0],X[:,1],c=y)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression(penalty=None,solver='sag')
lr.fit(X,y)
```

```python
m1=lr.coef_[0][0]/lr.coef_[0][1]
b1=lr.intercept_/lr.coef_[0][1]
```

```python
x_input=np.linspace(-3,3,100)
y_input1=m1*x_input+b1
```

```python
lr.intercept_

```

```python
lr.coef_
```

```python
def MyLogisticRegression(X,y):
    X=np.insert(X,0,1,axis=1)
    w=np.ones(X.shape[1])
    lr=0.9
    for i in range(2000):
        xw=np.dot(X,w)
        y_hat=1/(1+np.exp(-xw))
        w=w+lr*(np.dot((y-y_hat),X))/X.shape[0]
    return w[0],w[1:]
```

```python
amr=MyLogisticRegression(X,y)
```

```python
print(amr)
```

```python
inter=amr[0]
print(inter)
```

```python
coef=amr[1]
```

```python
print(coef)
```

```python
m=coef[0]/coef[1]
b=inter/coef[1]
```

```python
y_input=m*x_input+b
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y)
plt.plot(x_input,y_input)
plt.plot(x_input,y_input1)
plt.ylim(-3,2)
```

```python
def MyLogisticRegressionStochastic(X,y):
    X=np.insert(X,0,1,axis=1)
    w=np.ones(X.shape[1])
    lr=0.5
    for i in range(20):
        for j in range(X.shape[0]):
            idx=np.random.randint(0,X.shape[0])
            xw=np.dot(X[idx],w)
            y_hat=1/(1+np.exp(-xw))
            w=w+lr*(np.dot((y[idx]-y_hat),X[idx]))
    return w[0],w[1:]
```

```python
amrS=MyLogisticRegressionStochastic(X,y)
```

```python
print(amrS)
```

```python
inter2=amrS[0]
```

```python
coef2=amrS[1]
```

```python
m2=coef2[0]/coef2[1]
b2=inter2/coef2[1]
```

```python
y_input2=m*x_input+b
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y)
plt.plot(x_input,y_input)
plt.plot(x_input,y_input1)
plt.plot(x_input,y_input2)
plt.ylim(-3,2)
```
